```python
import ROOT
import h5py
import numpy as np
```

```python
Cal=h5py.File("../Outfile_MLPred_HomDet_15to20GeV_FineGran.h5","r")
```

```python
Cal.keys()
```

```python
lay=[64,64,32,16,16,8]
N_EV=len(Cal["Predicted_NeutralEnergy_Layer1"])
```

```python
Cal_lay=[]
Truth_lay=[]
for l in range(len(lay)):
    Cal_lay.append(np.array(Cal["Predicted_NeutralEnergy_Layer"+str(l+1)]))
    Truth_lay.append(np.array(Cal["Truth_NeutralEnergy_Layer"+str(l+1)]))
```

```python
h_pred=ROOT.TH1D("ENN","ENN",700, -100,1000)
h_truth=ROOT.TH1D("ENN","ENN",700, -100,1000)
for l in range(len(lay)):
    print("layer",l+1)
    for ev in range(N_EV):
        if ev%500==0: print("event",ev+1)
        for X in range(lay[l]):
            for Y in range(lay[l]):
                h_pred.Fill(Cal_lay[l][ev][0][X][Y])
                h_truth.Fill(Truth_lay[l][ev][0][X][Y])
```

```python
c = ROOT.TCanvas("","")
h_pred.SetLineColor(3)
h_pred.Draw()
h_truth.Draw("same")
```
