```python
import numpy as np
import h5py
import ROOT 
```

```python
f = h5py.File('../Outfile_CellInformation_HomDet_2to5GeV_FineGran.h5','r')
```

```python
f["RealRes_TotalEnergy_Layer"+str(6)].shape
```

```python
layers=[128,64,64,16,16,8]
```

```python
hist1=ROOT.TH1D("","",100,-100,100)
hist2=ROOT.TH1D("","",100,-100,100)
hist3=ROOT.TH1D("","",100,-100,100)
hist4=ROOT.TH1D("","",100,-100,100)
hist5=ROOT.TH1D("","",100,-100,100)
hist6=ROOT.TH1D("","",100,-100,100)


h=[hist1,hist2,hist3,hist4,hist5,hist6]

for lay in range(6):
    for ev in range(1000):
        if ev%100==0: print("lay",lay," loop ev: ", ev)

        T=f["RealRes_TotalEnergy_Layer"+str(lay+1)][ev][0]
        C=f["RealRes_ChargedEnergy_Layer"+str(lay+1)][ev][0]
        N=f["RealRes_NeutralEnergy_Layer"+str(lay+1)][ev][0]
        
        for X in range(layers[lay]):
            for Y in range(layers[lay]):
                #if T[X][Y]<0:
                #    print(X, Y, ev)
                h[lay].Fill(T[X][Y]-C[X][Y]-N[X][Y])
                #hist2.Fill(C[X][Y])
                #hist3.Fill(N[X][Y])
                #hist4.Fill(T[X][Y]-C[X][Y]-N[X][Y])
                

```

```python
can=ROOT.TCanvas( "","",1600,900);
can.Divide(2,3)
for lay in range(6):
    can.cd(lay)
    h[lay].SetLineColor(lay+1)
    h[lay].Draw("")
    can.Print("noise"+str(lay)+".png")
```
